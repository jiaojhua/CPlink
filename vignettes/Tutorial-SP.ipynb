# CPlink tutorial for spatial proteome data

Here, we provide a simplified and fast demonstration of **CPlink** on cancer CODEX data and bulk RNA-seq data that include both cancer and healthy samples. The bulk RNA-seq data were obtained from [TCGA](https://portal.gdc.cancer.gov/projects/tcga-luad) using TCGAbiolinks, and the spatial proteome data was obtained from [Takano et al](https://kero.hgc.jp/Ad-SpatialAnalysis_2024.html).

The processed data in matrix format also can be obtained from [here](https://drive.google.com/drive/folders/1BmtRkIVhDxlSkDKwQKWtVtS48Yo01eSc?usp=sharing)

```r
library(DESeq2)
library(Seurat)
library(ggplot2)
library(CPlink)
library(CellID)
set.seed(2025)
```

```r
setwd("~/SP/")
```

# Load preprocessed data

```r
bulk_count <- readRDS("count_matrix.rds")
bulk_meta <- readRDS("bulk_meta.rds")
codex_obj <- readRDS("codex_seurat.rds")
xenium_obj <- readRDS("xenium_obj.rds")
SP_coords <- readRDS("CODEX_coords.rds")
ST_coords <- readRDS("xenium_coords.rds")
```

# Identifying phenotype-specific genes from bulk data and getting cell-cell similarity graph from target data and data matrix from bridge data

```r
bulk_meta$phenotype <- relevel(bulk_meta$phenotype,ref='Normal')
Phe_Genes <- runDESeq2(bulk_count, bulk_meta, design_formula = ~ phenotype, p.adj = 0.05)
GEs_mtx <- data.frame(markers=rownames(Phe_Genes[[2]]), logFC=Phe_Genes[[2]]$log2FoldChange)
rownames(GEs_mtx) <- rownames(Phe_Genes[[2]])
```

```r
ST_data_matrix <- as.matrix(GetAssayData(xenium_obj, slot = "data"))
start <- Sys.time()
CODEX_nn <- getCOSNN(codex_obj, 0.2, SP_coords, snn = codex_obj@graphs$Akoya_snn)
end <- Sys.time()
end-start
```

# Run CPlink

Phenotype-associated signals are transferred to CODEX cells using the Xenium bridge.

```r
start <- Sys.time()
InitScore <- getCodexInitCellScore(ST_data_matrix, GEs_mtx, ST_coords = ST_coords, SP_coords = SP_coords)
CPlink_out_mtx <- runCPlink_CODEX(InitCellScore = InitScore, snn = CODEX_nn, seed_ratio = NULL, restart_prob = NULL, 
                            SigThreshold = 0.05, nCores = 16)
end <- Sys.time()
```

```r
end-start
```

# Attach CPlink results to Seurat meta.data

```r
seurat_obj <- codex_obj
seurat_obj@meta.data <- cbind(seurat_obj@meta.data, 
                                  cbind(CPlink_out_mtx[[1]], CPlink_out_mtx[[2]])[rownames(seurat_obj@meta.data),])
```
